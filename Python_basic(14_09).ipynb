2.**Pascal Case** : Each word starts with a capital letter:

3.**Snake Case**: Each word is separated by an underscore character:

Slicing Strings

```python
#start-inclusive, end-exclusive.

text = "I am monisha"

print(text[2:6])

#Slice From the Start
print(text[:6])

#Slice To the End
print(text[3:])

#Negative Indexing
print(text[-5:-2])
```

```python
#Modify Strings

str = "  Hello world  "

print(str.upper())

print(str.lower())

#The strip() method removes any whitespace from the beginning or the end:
print(str.strip())

#replace is case sensetive
print(str.replace('h','t'))
print(str.replace('H','t'))

print(str.split(" "))

text = "Hello , world"
print(text.split(","))
```

String Format (F-Strings)
 f means fixed-point notation.

```python
age = 20
price = 30
print(f"My age is {age}")

#{ variable : format_specifier }

print(f"Price of pen is {price:.3f}")

print(f"The price is dollars {price*95.89 : .2f}")
```

common formatting codes

.2f	2 decimal places	25.68

.3f	3 decimal places	25.679

.2e	Scientific notation	2.57e+01

.2%	Percentage	25.68%

,.2f	Comma + 2 decimals	1,234.56

```python
x = 1233445.4567

print(f"{x:.2f}")
print(f"{x:.2e}")
print(f"{x:.3%}")
print(f"{x:,.2f}")
```

Identity Operators

```python
"""
is - Checks if both variables point to the same object in memory
== - Checks if the values of both variables are equal

"""
x = ["apple", "banana"]
y = ["apple", "banana"]
z = x

print(x is z)
print(x is y)
print(x == y)
```
