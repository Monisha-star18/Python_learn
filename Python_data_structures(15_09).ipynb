Change a Range of Item Values


```python
newList = [10,20,30,40,50,60]
print(f"before the change : {newList}")

newList[1:3] = [30,20]
print(f"after the change : {newList}")

#If you insert more items than you replace,
# the new items will be inserted where you specified, and the remaining items will move accordingly

newList[3:5] = [50,40,30]
print(f"after the change : {newList}")


#If you insert less items than you replace,
#the new items will be inserted where you specified, and the remaining items will move accordingly:

newList[3:6] = [100]
print(f"after the change : {newList}")
```

List Comprehension

```python
#if i want to filter and store the elemnt in new array

fruits = ["apple","banana","cherry","kiwi","mango"]
newList =[]

for x in fruits:
  if "e" in x:
    newList.append(x)
print(fruits)
print(newList)
```

```python
#here the list comprehension comes in

#newlist = [expression | for item in iterable | if condition == True (if is not compulsory)]


fruits = ["banana","apple","kiwi","cherry","mango"]

newList = [x for x in fruits if "e" in x ]

print(newList)
```

Normal       → no arguments

Positional   → values based on position

Default      → parameter has a default value

Keyword      → parameter_name=value

*args        → any number of positional arguments


```python
# Normal function
def greet():
    print("Hello Monisha")

greet()
```

```python
#Function with one positional argument
def greet(name):
    print("Hello", name)

greet("Monisha")
```

```python
#Default argument
def greet(name="User"):
    print("Hello", name)

greet()
```

```python
#Keyword argument
def student(name, age):
    print("Name:", name)
    print("Age:", age)

student(age=22, name="Monisha")
```

```python
# Arbitrary arguments — *args
#*arg - > allow to pass multiple non key arguments
#**kwarg ->allow to pass multiple  keyword arguments
# * unpacking operator


#arg
#the arguments sent is been collected as a tuple

def add(*args):
  # print(type(args))  collected as a tuple
  total = 0
  for x in args :
    total +=x
  return total

print(add(10,20,30))
print(add(10,20,30,40,50))


#can use diff name than args like nums

def add(*nums):
  total = 0
  for x in nums :
    total +=x
  return total

print(add(10,20,30))
print(add(10,20,30,40,50))
```

```python
# ** kwargs - pass multiple keyword argumnet

def address ( **kwargs):

  for key in kwargs.keys():
    print(key , end = " , ")
  print()
  for value in kwargs.values():
    print(value , end = " ")
  print()
  for keys,values in kwargs.items():
    print(f"{keys} : {values}")

address( doorNo = 12 , stree = "nethaji nagar " , landmark = "sidco post" , state = "TN" , country = "IN" )
```
